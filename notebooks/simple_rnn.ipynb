```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,SimpleRNN,Embedding
from tensorflow.keras.datasets import imdb
import numpy as np
import tensorflow as tf
```

```python
## load the imdb datasets and spliting it in train and test
max_features = 10000
(x_train,y_train),(x_test,y_test) = imdb.load_data(num_words=max_features)
```

```python
print("training data shape" ,x_train.shape) ## training data
print("test_data_shape" ,x_test.shape) ## test data of the training
```

```python
max_len = 500
x_train = pad_sequences(x_train,maxlen=max_len)
x_test = pad_sequences(x_test,maxlen=max_len)
```

```python
x_train
```

## train our simple rnn

```python
model = Sequential()
model.add(Embedding(max_features, 128, input_shape=(max_len,)))

model.add(SimpleRNN(128, activation='tanh'))
model.add(Dense(1, activation='sigmoid'))
model.summary()
```

```python
model.compile(optimizer='adam'
                ,loss='binary_crossentropy'
                ,metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping

## create a instance of early stoping
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)
```

```python
## training our model on the  train and teat data with 100 epoch
model.fit(x_train,y_train,epochs=10,batch_size=32,
          validation_split =0.2,
          validation_data=(x_test,y_test),callbacks=[early_stopping])
```

```python
 ## evaluate the modeel
 model.evaluate(x_test,y_test)
```

```python
model.evaluate(x_train,y_train)
```

```python
## save the model
model.save("sample_rnn_imdb.h5")
```

```python
def decoded_reviews(encoded_review):
    return ' '.join([reveerse_word_index.get(i-3,'?') for i in encoded_review])
```

```python
def preprocess_text(text):
   words = text.lower().split()
   encoded_review =[word_index.get(word,2)+3 for word in words]
   padded_revies =  pad_sequences([encoded_review],maxlen=max_len)
   return padded_revies
```

```python
## prediction function
def prediction_sentiments(review):
  preprocess_input = preprocess_text(review)
  prediction = model.predict(preprocess_input)
  sentiment = 'positive' if prediction[0][0] > 0.5 else 'negative'
  return sentiment,prediction[0][0]
```
