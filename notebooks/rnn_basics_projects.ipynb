# Rnn project

```python
from tensorflow.keras.preprocessing.text  import one_hot
```

```python
sentences = [
    "The sun is bright.",
    "I like tea.",
    "She is happy.",
    "Dogs can run.",
    "It is cold.",
    "We are ready.",
    "He reads books.",
    "The sky is blue.",
    "Time goes fast.",
    "Birds can fly.",
    "I see you.",
    "Please sit down.",
    "The door is open.",
    "This is easy.",
    "She sings well.",
    "The baby sleeps.",
    "Water is clear.",
    "I found it.",
    "The cat jumps.",
    "We love music."
]
```

```python
## we need to define vocablary size
```

```python
voc_size = 10000
```

```python
ohe_rep = [one_hot(words,voc_size)for words in sentences]
ohe_rep
```

## sentences words is got converted in vector with in the 10000 vocabllary

# Our each words got an index in the vocablery size out of 1000 like show in above list of vectors

## now we have a option to rduce the dim use our pca(principle components analyis)

# word embeding layers

## We pass our index vector to the embeding layers

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np
```

```python
sentences_len = 7
embedded_docs = pad_sequences(ohe_rep,padding='pre',maxlen=sentences_len)
embedded_docs
```

```python
# feature representation
dim = 10
```

```python
model = Sequential()
model.add(Embedding(voc_size, dim, input_shape=(sentences_len,)))
model.add(Dense(1))

model.compile(optimizer='adam', loss='mse')
model.summary()
```

```python
embedded_docs[0]
```

```python
model.predict(embedded_docs)
```
